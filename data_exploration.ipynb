# Data Exploration — Social Media Phishing Detection

This notebook explores the `sms.tsv` dataset used to train the phishing detection model.

**Note on dataset re-purposing:** We use the public SMS Spam Collection dataset as a stand-in for social-media phishing messages. `spam` messages are treated as **phishing** (label = 1) and `ham` messages are treated as **legitimate social media messages** (label = 0). This is documented in the README.

```python
import sys
sys.path.append('..')

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from utils import load_dataset
from preprocessing import clean_corpus

sns.set_style('whitegrid')
```

```python
df = load_dataset('../dataset/sms.tsv')
print(f'Total messages: {len(df)}')
df.head(10)
```

## Class distribution

```python
counts = df['label'].value_counts()
print(counts)

plt.figure(figsize=(5, 4))
sns.barplot(x=counts.index, y=counts.values, palette=['#4ADE80', '#FB6B6B'])
plt.title('Class Distribution: Legitimate (ham) vs Phishing (spam)')
plt.ylabel('Count')
plt.xlabel('Label')
plt.show()
```

## Message length analysis

```python
df['length'] = df['text'].apply(len)

plt.figure(figsize=(7, 4))
sns.histplot(data=df, x='length', hue='label', bins=40, kde=True, palette=['#4ADE80', '#FB6B6B'])
plt.title('Message Length Distribution by Class')
plt.xlabel('Message length (characters)')
plt.show()

df.groupby('label')['length'].describe()
```

## Preview cleaned text

```python
df['clean_text'] = clean_corpus(df['text'].tolist())
df[['label', 'text', 'clean_text']].sample(10, random_state=1)
```

## Most frequent words per class

```python
from collections import Counter

def top_words(subset_df, n=15):
    words = ' '.join(subset_df['clean_text']).split()
    return Counter(words).most_common(n)

print('Top words — Phishing (spam):')
print(top_words(df[df['label'] == 'spam']))

print('\nTop words — Legitimate (ham):')
print(top_words(df[df['label'] == 'ham']))
```
